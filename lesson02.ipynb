# Lesson 2 — Loading & inspecting data
**Goal:** load a year of real-looking sales (`orders.csv`, 2,400 orders) and get to know it in 60 seconds with `head`, `info`, `describe`, `value_counts`.

```python
import pandas as pd
orders = pd.read_csv('data/orders.csv')
orders.head()
```

```python
orders.shape
```

```python
orders.info()
```

Notice `order_date` loaded as text. Tell pandas it's a date with `parse_dates`:

```python
orders = pd.read_csv('data/orders.csv', parse_dates=['order_date'])
orders['order_date'].min(), orders['order_date'].max()
```

```python
orders.describe()   # quick stats for numeric columns
```

```python
orders['branch'].value_counts()   # how many orders per branch?
```

```python
orders['payment_method'].value_counts(normalize=True).round(3)  # as shares
```

## Other files, same idea
Excel, JSON, SQL all have a `read_*` function. Let's load the other tables we'll use:

```python
products  = pd.read_csv('data/products.csv')
customers = pd.read_csv('data/customers.csv', parse_dates=['signup_date'])
mm        = pd.read_csv('data/mobile_money.csv', parse_dates=['timestamp'])
for name, df in [('products', products), ('customers', customers), ('mobile money', mm)]:
    print(f'{name:13s} {df.shape}')
```

```python
# Reading Excel works the same way (needs openpyxl):
products.to_excel('data/products.xlsx', index=False)
pd.read_excel('data/products.xlsx').head(3)
```

## Key takeaways
- `pd.read_csv(path, parse_dates=[...])` — always parse your date columns.
- First 4 commands on any new dataset: `head()`, `info()`, `describe()`, `value_counts()`.
