# Lesson 1 — What is pandas?
**Goal:** understand the two core objects — the `Series` (one column) and the `DataFrame` (a table) — using a tiny sales table from our shop, **Duka Bora**.

> Real-world angle: every spreadsheet you've used is a DataFrame. pandas lets you do with code what you do by hand in Excel — but repeatable, and on millions of rows.

```python
import pandas as pd
import numpy as np
pd.__version__
```

## A Series = one labelled column
Think: the *Price* column of a price list. Values plus an **index** (the labels).

```python
prices = pd.Series([25000, 45000, 85000, 18000],
                   index=['Wireless Mouse', 'USB-C Charger', 'Bluetooth Earbuds', 'Flash Drive'],
                   name='price_tzs')
prices
```

```python
# Look up by label, do maths on the whole column at once (no loops!)
print(prices['Bluetooth Earbuds'])
print(prices.mean())
prices * 1.18   # add 18% VAT to every price
```

## A DataFrame = a table of Series that share an index
Each column is a Series. Here are 5 orders from today:

```python
orders = pd.DataFrame({
    'order_id': ['O001', 'O002', 'O003', 'O004', 'O005'],
    'product':  ['Wireless Mouse', 'Smartphone 128GB', 'Flash Drive', 'Bluetooth Earbuds', 'Wireless Mouse'],
    'quantity': [2, 1, 4, 1, 1],
    'unit_price': [25000, 650000, 18000, 85000, 25000],
    'branch':   ['Kariakoo', 'Mlimani City', 'Kariakoo', 'Arusha CBD', 'Mwanza Rock City'],
})
orders
```

```python
orders.shape   # (rows, columns)
```

```python
orders.dtypes  # the type of each column
```

## Your first calculation: revenue per order
Multiply two columns → a new column. pandas lines them up row by row.

```python
orders['revenue'] = orders['quantity'] * orders['unit_price']
orders
```

```python
print('Total revenue today: TSh', f"{orders['revenue'].sum():,}")
print('Best-selling order:', orders.loc[orders['revenue'].idxmax(), 'product'])
```

## Key takeaways
- `Series` = 1-D labelled array; `DataFrame` = 2-D table of Series.
- Operations apply to **whole columns** (vectorised) — no `for` loops.
- `shape`, `dtypes` tell you what you're working with.
